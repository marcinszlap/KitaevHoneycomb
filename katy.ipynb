```python
import numpy as np
from fractions import Fraction

def format_as_pi_fraction(value, max_denominator=12):
    """
    Zamienia liczbę (kąt / π) na string typu 'π/2', '2π/3', '0', itd.
    """
    frac = Fraction(value).limit_denominator(max_denominator)
    num, den = frac.numerator, frac.denominator

    if num == 0:
        return "0"
    
    sign = "-" if num < 0 else ""
    num = abs(num)

    if den == 1:
        if num == 1:
            return f"{sign}π"
        else:
            return f"{sign}{num}π"
    else:
        if num == 1:
            return f"{sign}π/{den}"
        else:
            return f"{sign}{num}π/{den}"


def cartesian_to_spherical_pi_fraction(x, y, z, max_denominator=12):
    r = np.sqrt(x**2 + y**2 + z**2)
    
    if r == 0:
        return 0.0, "0", "0"
    
    theta = np.arccos(z / r)
    phi = np.arctan2(y, x)
    
    theta_pi = theta / np.pi
    phi_pi = phi / np.pi
    
    theta_str = format_as_pi_fraction(theta_pi, max_denominator)
    phi_str = format_as_pi_fraction(phi_pi, max_denominator)
    
    return r, theta_str, phi_str

print(cartesian_to_spherical_pi_fraction(1, 1, 1))
```
